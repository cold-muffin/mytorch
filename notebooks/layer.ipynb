```python
import numpy as np
```

```python
NUM_FEATURES = None
NUM_SAMPLES = None
NUM_NEURONS_1 = None
NUM_NEURONS_2 = None
```

```python
num_features = NUM_FEATURES or 3
num_samples = NUM_SAMPLES or 4
inputs = np.random.rand(num_samples, num_features)
inputs
```

```python
num_neurons = NUM_NEURONS_1 or 5
weights = np.random.rand(num_neurons, num_features)
weights
```

```python
biases = np.random.rand(1, num_neurons)
biases
```

```python
output = inputs @ weights.T + biases
output
```

```python
inputs2 = output # output from previous layer becomes input for new layer
```

```python
num_neurons_2 = NUM_NEURONS_2 or 6 # number of neurons in second layer (hidden layer)
weights2 = np.random.rand(num_neurons_2, num_neurons)
weights2
```

```python
biases2 = np.random.rand(1, num_neurons_2)
biases2
```

```python
output2 = inputs2 @ weights2.T + biases2
output2
```
