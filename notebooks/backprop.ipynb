```python
import numpy as np
from dense_layer import DenseLayer
```

```python
layer0 = DenseLayer(3, 2)
layer1 = DenseLayer(2, 1)
inp = np.random.random((1, 3))
inp
```

```python
def forward(inp: np.ndarray) -> np.ndarray:
  global layer0
  global layer1
  return layer1.forward(layer0.forward(inp))
```

```python
forward(inp)
```

```python
def backward(num_epochs: int):
  global layer0
  global layer1
  for epoch in range(num_epochs):
    layer0.backprop(layer1.backprop())
```

```python
backward(1000000)
forward(inp)
```
